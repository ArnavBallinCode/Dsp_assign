```python
# Iris Species Dataset: https://www.kaggle.com/datasets/uciml/iris/data 
 
# 1.  Create a figure with two subplots side by side: 
 
# ○  Left subplot: 
# ■  Plot a scatter plot of sepal_length vs sepal_width. 
# ■  Use different colors for each species. 
# ■  Add a legend with species names. 
# ■  Set custom x-ticks and rotate them. 
 
# ○  Right subplot: 
# ■  Plot a histogram of petal_length (different colors for species). 
# ■  Add labels, title, and legend. 
 
# 2.  Customize both plots: 
 
# ○  Add titles, x/y-axis labels. 
# ○  Use ax.set_xticks(), ax.set_xticklabels(rotation=45, fontsize=10). 
# ○  Use fig.subplots_adjust(wspace=0.3) to control spacing. 
 
# 3.  Save the figure as "iris_visualization.png" with dpi=300. 


import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)

# Display the first few rows of the dataframe
data.head()
```

```python
# 1.  Create a figure with two subplots side by side: 
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# ○  Left subplot: 
# ■  Plot a scatter plot of sepal_length vs sepal_width. 
# ■  Use different colors for each species. 
# ■  Add a legend with species names. 
species = data['species'].unique()
colors = ['r', 'g', 'b']
for i, sp in enumerate(species):
    subset = data[data['species'] == sp]
    axs[0].scatter(subset['sepal_length'], subset['sepal_width'], color=colors[i], label=sp)
axs[0].set_xlabel('Sepal Length')
axs[0].set_ylabel('Sepal Width')
axs[0].set_title('Sepal Length vs Width')
axs[0].legend()
# ■  Set custom x-ticks and rotate them.
axs[0].set_xticks(np.arange(4, 8, 0.5))
axs[0].set_xticklabels(np.arange(4, 8, 0.5), rotation=45, fontsize=10)


# ○  Right subplot: 
# ■  Plot a histogram of petal_length (different colors for species). 
# ■  Add labels, title, and legend. 
for i, sp in enumerate(species):
    subset = data[data['species'] == sp]
    axs[1].hist(subset['petal_length'], color=colors[i], alpha=0.5, label=sp, bins=10)
axs[1].set_xlabel('Petal Length')
axs[1].set_ylabel('Frequency')
axs[1].set_title('Petal Length Distribution')
axs[1].legend()
```

```python
# 2.  Customize both plots: 

# ○  Add titles, x/y-axis labels.
axs[0].set_xlabel('Sepal Length')
axs[0].set_ylabel('Sepal Width')
axs[0].set_title('Sepal Length vs Width')

axs[1].set_xlabel('Petal Length')
axs[1].set_ylabel('Frequency')
axs[1].set_title('Petal Length Distribution')

# ○  Use ax.set_xticks(), ax.set_xticklabels(rotation=45, fontsize=10).
axs[0].set_xticks(np.arange(4, 8, 0.5))
axs[0].set_xticklabels(np.arange(4, 8, 0.5), rotation=45, fontsize=10)

# ○  Use fig.subplots_adjust(wspace=0.3) to control spacing.
fig.subplots_adjust(wspace=0.3)


# 3.  Save the figure as "iris_visualization.png" with dpi=300. 

plt.savefig("iris_visualization.png", dpi=300)
```
