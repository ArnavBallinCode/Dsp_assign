```python
# import pandas numpy and matplotlib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load the titanic assignment 
titanic_data = pd.read_csv('titanic.csv')

# Plot how the number of passengers varies by age using a line plot
plt.plot(titanic_data['Age'], titanic_data['PassengerId'], color='g', marker='o')

# Add xlabel, ylabel, and title.
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.title('Number of Passengers by Age')

```

```python
# Create two separate figures (plt.figure()) 
#  ○  First figure: Fare vs PassengerId. 
#  ○  Second figure: Age vs PassengerID

plt.figure()
plt.plot(titanic_data['Age'], titanic_data['PassengerId'], color='g', marker='o')
plt.xlabel('Age')
plt.ylabel('PassengerID')
plt.title('PassengerID by Age')

plt.figure()
plt.plot(titanic_data['Fare'], titanic_data['PassengerId'], color='b', marker='o')
plt.xlabel('Fare')
plt.ylabel('PassengerID')
plt.title('PassengerID by Fare')
```

```python
# Use fig.add_subplot(2,2,1) style to create a 2x2 grid:
# ○ Top-left: Histogram of Age.
# ○ Top-right: Histogram of Fare.
# ○ Bottom-left: Line plot of Age vs Fare (random 100 samples).
# ○ Bottom-right: Empty for practice (let them choose any visualization).





fig = plt.figure(figsize=(12, 10))

# Top-left: Histogram of Age
ax1 = fig.add_subplot(2, 2, 1)
ax1.hist(titanic_data['Age'].dropna(), bins=20, color='skyblue', edgecolor='black')
ax1.set_title('Histogram of Age')
ax1.set_xlabel('Age')
ax1.set_ylabel('Frequency')

# Top-right: Histogram of Fare
ax2 = fig.add_subplot(2, 2, 2)
ax2.hist(titanic_data['Fare'], bins=20, color='salmon', edgecolor='black')
ax2.set_title('Histogram of Fare')
ax2.set_xlabel('Fare')
ax2.set_ylabel('Frequency')

# Bottom-left: Line plot of Age vs Fare (random 100 samples)
ax3 = fig.add_subplot(2, 2, 3)
sample = titanic_data[['Age', 'Fare']].dropna().sample(100, random_state=1)
ax3.plot(sample['Age'], sample['Fare'], marker='o', linestyle='-', color='green')
ax3.set_title('Age vs Fare (100 random samples)')
ax3.set_xlabel('Age')
ax3.set_ylabel('Fare')

# Bottom-right: Empty subplot for practice
ax4 = fig.add_subplot(2, 2, 4)
ax4.set_title('Your Visualization Here')
ax4.axis('off')





plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(12, 10))
ax4 = fig.add_subplot(2, 2, 4)


# Plot histogram of Age
ax4.hist(titanic_data['Age'].dropna(), bins=20, color='k', alpha=0.5, label='Age')

# Plot histogram of Fare on the same figure
ax4.hist(titanic_data['Fare'], bins=20, color='orange', alpha=0.5, label='Fare')

ax4.set_title('Histogram of Age and Fare')
ax4.set_xlabel('Value')
ax4.set_ylabel('Frequency')
ax4.legend()
fig.tight_layout()
fig.canvas.draw()
plt.show()
```
