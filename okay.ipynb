```python
import matplotlib.pyplot as plt
import numpy as np

theta = np.linspace(0, 2 * np.pi, 100)  # Angle values
r = np.abs(np.sin(theta))  # Radius values

plt.polar(theta, r)
plt.title("Polar Plot of r = |sin(θ)|")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

# all subplots
plt.subplot(2, 1, 1)
plt.plot(x, y)
plt.title("Plot of y = sin(x)")

plt.subplot(2, 1, 2)
plt.plot(x, y**2)
plt.title("Plot of y = sin^2(x)")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  

# different labels and different colours, graph ofd sinx cosx and tanx with a proper y axis and stuff, so all graphs are clear 
x = np.linspace(0,2*np.pi,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

plt.plot(x,y1,label='sin(x)',color='blue')
plt.plot(x,y2,label='cos(x)',color='red')
plt.plot(x,y3,label='tan(x)',color='green')
plt.legend()

# save the png using all different arguments , dpu and bbox inches tithgt and all
plt.savefig("trig_functions.png", dpi=300, bbox_inches='tight')
```
