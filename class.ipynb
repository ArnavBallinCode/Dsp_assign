```python
import numpy as np

import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.exp(-x/3) * np.sin(2*x)
y4 = np.random.rand(100)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
axs[0, 0].plot(x, y1, color='blue' ,)
axs[0, 0].set_title('Line Plot (sin(x))')

# Scatter plot
axs[0, 1].scatter(x, y2, color='red')
axs[0, 1].set_title('Scatter Plot (cos(x))')

# Bar plot
axs[1, 0].bar(x[::10], y3[::10], color='green')
axs[1, 0].set_title('Bar Plot (exp(-x/3)*sin(2x))')

# Histogram
axs[1, 1].hist(y4, bins=10, color='purple')
axs[1, 1].set_title('Histogram (random data)')

plt.tight_layout()
plt.show()
```
