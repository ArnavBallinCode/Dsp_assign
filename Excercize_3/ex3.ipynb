```python
import pandas as pd

# Expanded mapping
rating_map = {
    "Excellent": 5,
    "Very Good": 4,
    "Good": 3,
    "Average": 2,
    "Poor": 1,
    "Strongly agree": 5,
    "Agree": 4,
    "Neutral": 3,
    "Disagree": 2,
    "Strongly disagree": 1,
    "Satisfactory": 3
}

# Load CSV
df = pd.read_csv("feedback.csv")

# Columns to ignore (non-rating text)
exclude_cols = [
    "Timestamp",
    "What aspects of this course were most useful or valuable?",
    "How would you improve this course?",
    "Any other comments"
]

# Rating columns = everything except excluded
rating_columns = [col for col in df.columns if col not in exclude_cols]

# Apply mapping
for col in rating_columns:
    df[col] = df[col].map(rating_map)

#  Filter rows where *any rating column* >= 4
filtered_df = df[df[rating_columns].ge(4).any(axis=1)]

# Save filtered feedback
filtered_df.to_csv("filtered_feedback.csv", index=False)

# Calculate averages
overall_avg = df[rating_columns].mean().mean()
per_column_avg = df[rating_columns].mean()

print("Filtered feedback saved to filtered_feedback.csv")
print(f"Overall Average Rating: {overall_avg:.2f}")
print("\nAverage per Question:\n", per_column_avg)
```
